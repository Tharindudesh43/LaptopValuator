```python
import numpy as np
import pandas as pd
```

```python
#load csv file
data = pd.read_csv("laptop_price.csv",encoding='latin-1')
data.head()
```

```python
data.shape
```

```python
#check null values
data.isnull().sum()
```

```python
data.info()
```

```python
data.head(2)
```

```python
data['Ram'] = data['Ram'].str.replace('GB','').astype('int32')
data['Weight'] = data['Weight'].str.replace('kg','').astype('float32')
```

```python
#find most resonable columns for price_euros data chnages
data.corr(numeric_only=True)['Price_euros']
```

```python
data['Company'].value_counts()
```

```python
def add_company(input):
    if input=='Samsung' or input=='Razer' or input=='Mediacom' or input=='Microsoft' or input=='Xiaomi' or input=='Vero' or input=='Chuwi' or input=='Google' or input=='Fujitsu' or input=='LG' or input=='Huawei':
        return 'Other'
    else:
        return input

data['Company'] = data['Company'].apply(add_company)
```

```python
data['Company'].value_counts()
```

```python
len(data['Product'].value_counts())
```

```python
data['TypeName'].value_counts()
```

```python
data['ScreenResolution'].value_counts()
```

```python
data['Touchscreen'] = data['ScreenResolution'].apply(lambda x:1 if 'Touchscreen' in x else 0)
data['Ips'] = data['ScreenResolution'].apply(lambda x:1 if 'IPS' in x else 0)
```

```python
 data.head(2)
```

```python
len(data['Cpu'].value_counts())
```

```python
data['cpu_name'] = data['Cpu'].apply(lambda x:" ".join(x.split()[0:3]))
```

```python
data['cpu_name'].value_counts()
```

```python
def  set_processor(name):
    if name== 'Intel Core i7' or name== 'Intel Core i5' or name== 'Intel Core i3':
        return name
    else:
        if name.split()[0] == 'AMD':
            return 'AMD'
        else:
            'Other'

data['cpu_name'] = data['cpu_name'].apply(set_processor)
```

```python
data['cpu_name'].value_counts()
```

```python
data['Gpu'].value_counts()
```

```python
data['gpu_name'] = data['Gpu'].apply(lambda x:" ".join(x.split()[0:1]))
```

```python
data['gpu_name'].value_counts()
```

```python
data.shape
```

```python
data = data[data['gpu_name'] is 'ARM']
```

```python
data.shape
```

```python
data.head(2)
```

```python
data['OpSys'].value_counts()
```

```python
#reduce categories
def set_os(input):
    if input == 'Windows 10' or input == 'Windows 7' or input == 'Windows 10 S':
        return 'Windows'
    elif input == 'macOS' or input == 'Mac OS X':
        return 'Mac'
    elif input == 'Linux':
        return input
    else:
        return 'Other'

data['OpSys'] = data['OpSys'].apply(set_os)
```

```python
data['OpSys'].value_counts()
```

```python
data.head(2)
```

```python
#drop those columns from data
data = data.drop(columns=['laptop_ID','Inches','Product','ScreenResolution','Cpu','Gpu'])
```

```python
data.head(2)
```

```python
#one hot encoding
data = pd.get_dummies(data)
```
